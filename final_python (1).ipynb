```python
# Import Libraries & Load Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r'C:\Users\Sarvadnya\Downloads\HR_Data.csv')
```

```python
# Q1. Identify Missing Values
df.isnull().sum()
```

```python
# Q2. Check Duplicate Records
df.duplicated(subset='Employee_ID').sum()
df = df.drop_duplicates(subset='Employee_ID')
df
```

```python
# 3. Convert Hire_Date & Create Tenure
df['Hire_Date'] = pd.to_datetime(df['Hire_Date'])
df
```

```python
df['Tenure_Years'] = (pd.Timestamp('today') - df['Hire_Date']).dt.days / 365
df
```

```python
# Q4. Experience > Tenure (Inconsistency)
```

```python
mask = df['Experience_Years'] > df['Tenure_Years']
df.loc[mask, 'Experience_Years'] = df.loc[mask, 'Tenure_Years']
```

```python
mask
```

```python
df
```

```python
df[['Department', 'Job_Title']].dtypes
```

```python
df['Department'] = df['Department'].astype(str)
df['Job_Title'] = df['Job_Title'].astype(str)
```

```python
# 5. Standardize Department & Job Title
df['Department'] = df['Department'].str.strip().str.upper()
df['Job_Title'] = df['Job_Title'].str.strip().str.title()
df
```

```python
# Why important? Avoids duplicate labels in dashboards ("IT", "it", " It").
```

```python
# Q6. Check Negative or Zero Salary
df[df['Salary_INR']<=0]
```

```python
# Impact: Breaks salary budgeting, payroll analysis, and cost forecasting.
```

```python
# Q7. Salary Outliers (Percentile)
```

```python
low = np.percentile(df['Salary_INR'],1)
high = np.percentile(df['Salary_INR'],99)
```

```python
outliers = df[(df['Salary_INR'] < low) | (df['Salary_INR'] > high)]
outliers
```

```python
# insight: Outliers may be CXO salaries (business insight) or data entry errors.
```
